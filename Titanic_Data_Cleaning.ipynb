```python

import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python

df.isnull().sum()
```

```python

cleaned_df = df.copy()

print("Original dataset shape:", df.shape)
print("Working copy shape:", cleaned_df.shape)
```

```python

cleaned_df["age"] = cleaned_df["age"].fillna(
    cleaned_df["age"].median()
)

print("Missing age values after cleaning:")
print(cleaned_df["age"].isnull().sum())
```

```python

print("Missing embarked values before cleaning:")
print(cleaned_df["embarked"].isnull().sum())

print("Most frequent embarked value:")
print(cleaned_df["embarked"].mode()[0])
```

```python

cleaned_df["embarked"] = cleaned_df["embarked"].fillna(
    cleaned_df["embarked"].mode()[0]
)

print("Missing embarked values after cleaning:")
print(cleaned_df["embarked"].isnull().sum())
```

```python

print("Missing embark_town values before cleaning:")
print(cleaned_df["embark_town"].isnull().sum())

print("\nMost frequent embark_town value:")
print(cleaned_df["embark_town"].mode()[0])
```

```python

cleaned_df["embark_town"] = cleaned_df["embark_town"].fillna(
    cleaned_df["embark_town"].mode()[0]
)

print("Missing embark_town values after cleaning:")
print(cleaned_df["embark_town"].isnull().sum())
```

```python

print("Missing deck values before cleaning:")
print(cleaned_df["deck"].isnull().sum())
```

```python

cleaned_df["deck"] = cleaned_df["deck"].fillna("Unknown")

print("Missing deck values after cleaning:")
print(cleaned_df["deck"].isnull().sum())

print("\nUnknown deck values:")
print((cleaned_df["deck"] == "Unknown").sum())
```

```python

print("Missing values in each column:")
print(cleaned_df.isnull().sum())

print("\nTotal missing values remaining:")
print(cleaned_df.isnull().sum().sum())
```

```python

print("Duplicate records before removal:")
print(cleaned_df.duplicated().sum())
```

```python

cleaned_df = cleaned_df.drop_duplicates()

print("Duplicate records after removal:")
print(cleaned_df.duplicated().sum())

print("\nDataset shape after removing duplicates:")
print(cleaned_df.shape)
```

```python

print("Remaining duplicate records:",
      cleaned_df.duplicated().sum())

print("Rows after cleaning:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
```

```python

print("Data types before correction:")
print(cleaned_df.dtypes)
```

```python

for col in ["survived", "pclass", "sibsp", "parch"]:
    cleaned_df[col] = cleaned_df[col].astype("int64")

for col in ["age", "fare"]:
    cleaned_df[col] = pd.to_numeric(cleaned_df[col], errors="coerce")

text_cols = [
    "sex", "embarked", "class", "who",
    "deck", "embark_town", "alive"
]

for col in text_cols:
    cleaned_df[col] = cleaned_df[col].astype("string")

for col in ["adult_male", "alone"]:
    cleaned_df[col] = cleaned_df[col].astype("bool")

print("Data types after correction:")
print(cleaned_df.dtypes)
```

```python

print("Data type validation:")

print("Integer columns:",
      all(cleaned_df[c].dtype == "int64"
          for c in ["survived", "pclass", "sibsp", "parch"]))

print("Numeric columns:",
      all(pd.api.types.is_numeric_dtype(cleaned_df[c])
          for c in ["age", "fare"]))

print("Text columns:",
      all(cleaned_df[c].dtype == "string"
          for c in text_cols))

print("Boolean columns:",
      all(cleaned_df[c].dtype == "bool"
          for c in ["adult_male", "alone"]))
```

```python

text_cols = [
    "sex", "embarked", "class", "who",
    "deck", "embark_town", "alive"
]

print("Checking text columns for extra spaces:\n")

for col in text_cols:
    spaces = cleaned_df[col].str.strip().ne(cleaned_df[col]).sum()
    print(col, ":", spaces, "values with extra spaces")

print("\nUnique values in each text column:\n")

for col in text_cols:
    print(col, ":", cleaned_df[col].unique())
```

```python

for col in text_cols:
    cleaned_df[col] = cleaned_df[col].str.strip()

print("Text values standardized successfully!")

for col in text_cols:
    print(col, ":", cleaned_df[col].nunique(), "unique values")
```

```python

print("FINAL DATA CLEANING REPORT")
print("--------------------------")

print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Total missing values:", cleaned_df.isnull().sum().sum())
print("Duplicate records:", cleaned_df.duplicated().sum())

print("\nMissing values by column:")
print(cleaned_df.isnull().sum())

print("\nFinal validation completed!")
```

```python

cleaned_df.to_csv("titanic_cleaned.csv", index=False)

from google.colab import files
files.download("titanic_cleaned.csv")

```
